```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
```

```python
X = tf.range(10)
```

```python
X
```

```python
dataset = tf.data.Dataset.from_tensor_slices(X)
```

```python
for item in dataset:
    print(item)
```

```python
dataset = dataset.repeat(3).batch(7)
for item in dataset:
    print(item)
```

```python
dataset = dataset.map(lambda x:x*2)
for i in dataset:
    print(i)
```

```python
dataset = dataset.apply(tf.data.experimental.unbatch())
```

```python
new_dataset = tf.range(10)
```

```python
new_dataset = tf.data.Dataset.from_tensor_slices(new_dataset)
```

```python
new_dataset = new_dataset.filter(lambda x:x<5)
```

```python
for i in new_dataset:
    print(i)
```

```python
for i in new_dataset.take(1):
    print(i)
```

```python
n_ds = tf.data.Dataset.range(10).repeat(3)
```

```python
n_ds = n_ds.shuffle(buffer_size=5, seed=42).batch(7)
```

```python
for i in n_ds:
    print(i)
```
